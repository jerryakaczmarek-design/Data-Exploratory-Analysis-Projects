## 5) Analysis and Visualizations

###Visualization 1: Average price of avacado over time that highlights changes in prices over time

```python
price_over_time = df_csv.groupby('Date')['AveragePrice'].mean().reset_index() # Group by Date and calculate the mean of AveragePrice for the line plot

plt.figure(figsize=(15, 7))
sns.lineplot(x='Date', y='AveragePrice', data=price_over_time)
plt.title('Average Avocado Price Over Time (avocado.csv)')
plt.xlabel('Date')
plt.ylabel('Average Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Visualization 2: Average avacado price comparison organic vs conventional

```python
price_by_type_date = df_csv.groupby(['Date', 'type'])['AveragePrice'].mean().reset_index() # Group by Date and type, then calculate the mean of AveragePrice

plt.figure(figsize=(15, 7))
sns.lineplot(x='Date', y='AveragePrice', hue='type', data=price_by_type_date)
plt.title('Average Avocado Price Over Time by Type (Organic vs. Conventional) (avocado.csv)')
plt.xlabel('Date')
plt.ylabel('Average Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.legend(title='Avocado Type')
plt.tight_layout()
plt.show()
```

### Visualization 3: Average price of avacados across regions

```python
average_price_by_region = df_csv.groupby('region')['AveragePrice'].mean().reset_index() # Calculate the average price for each region

average_price_by_region = average_price_by_region.sort_values(by='AveragePrice', ascending=False) # Sort the regions by average price for better visualization

plt.figure(figsize=(16, 8))
sns.barplot(x='region', y='AveragePrice', hue='region', data=average_price_by_region, palette='viridis', legend=False)
plt.title('Average Avocado Price Across Different Regions (avocado.csv)')
plt.xlabel('Region')
plt.ylabel('Average Price')
plt.xticks(rotation=90) # Rotate x-axis labels for readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

### Visualization 4L Average avacado price by season

```python
# Function to map month to season for visualization
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'

# Create a 'Season' column
df_csv['Season'] = df_csv['Date'].dt.month.apply(get_season)

# Calculate the average price per season
average_price_by_season = df_csv.groupby('Season')['AveragePrice'].mean().reset_index()

# Define a specific order for seasons for better visualization
season_order = ['Spring', 'Summer', 'Fall', 'Winter']
average_price_by_season['Season'] = pd.Categorical(average_price_by_season['Season'], categories=season_order, ordered=True)
average_price_by_season = average_price_by_season.sort_values('Season')

plt.figure(figsize=(10, 6))
sns.barplot(x='Season', y='AveragePrice', hue='Season', data=average_price_by_season, palette='coolwarm', legend=False)
plt.title('Average Avocado Price by Season (avocado.csv)')
plt.xlabel('Season')
plt.ylabel('Average Price')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Visualization 5: Total avacado sales by month

```python
# Extract month from the 'Date' column in df_csv
df_csv['month'] = df_csv['Date'].dt.month

# Group by month and sum the 'Total Volume' for df_csv
monthly_sales_csv = df_csv.groupby('month')['Total Volume'].sum().reset_index()

# Create a bar plot for total sales volume by month from df_csv
plt.figure(figsize=(12, 6))
sns.barplot(x='month', y='Total Volume', hue='month', data=monthly_sales_csv, palette='crest', legend=False)
plt.title('Total Avocado Sales Volume by Month (avocado.csv)')
plt.xlabel('Month')
plt.ylabel('Total Volume')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
